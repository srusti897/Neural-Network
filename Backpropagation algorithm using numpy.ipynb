1.(a) Backpropagation Algorithm Using Numpy

```python
import numpy as np   # Import NumPy for numerical operations
# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))   # Maps values between 0 and 1
# Derivative of sigmoid function (used during backpropagation)
def sigmoid_derivative(x):
    return x * (1 - x)
# Input dataset (XOR inputs)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
# Expected output
y = np.array([[0], [1], [1], [0]])
# Set random seed for reproducibility
np.random.seed(42)
# Number of neurons in each layer
input_layer_neurons = 2
hidden_layer_neurons = 2
output_neurons = 1
# Initialize weights randomly
weights_input_hidden = np.random.uniform(size=(input_layer_neurons, hidden_layer_neurons))
weights_hidden_output = np.random.uniform(size=(hidden_layer_neurons, output_neurons))
# Initialize bias values
bias_hidden = np.random.uniform(size=(1, hidden_layer_neurons))
bias_output = np.random.uniform(size=(1, output_neurons))
# Learning rate controls speed of weight updates
learning_rate = 0.1
# Training the neural network
for epoch in range(10000):
    # -------- FORWARD PROPAGATION --------
    # Compute input to hidden layer
    hidden_input = np.dot(X, weights_input_hidden) + bias_hidden
    # Apply activation function
    hidden_output = sigmoid(hidden_input)
    # Compute input to output layer
    final_input = np.dot(hidden_output, weights_hidden_output) + bias_output
    # Apply activation function to get final output
    predicted_output = sigmoid(final_input)
    # -------- BACKWARD PROPAGATION --------
    # Calculate error between actual and predicted output
    error = y - predicted_output
    # Gradient of output layer
    d_output = error * sigmoid_derivative(predicted_output)
    # Error propagated back to hidden layer
    error_hidden = d_output.dot(weights_hidden_output.T)
    # Gradient of hidden layer
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)
    # -------- UPDATE WEIGHTS AND BIASES --------
    # Update weights between hidden and output layer
    weights_hidden_output += hidden_output.T.dot(d_output) * learning_rate
    # Update weights between input and hidden layer
    weights_input_hidden += X.T.dot(d_hidden) * learning_rate
    # Update bias of output layer
    bias_output += np.sum(d_output, axis=0, keepdims=True) * learning_rate
    # Update bias of hidden layer
    bias_hidden += np.sum(d_hidden, axis=0, keepdims=True) * learning_rate
# Display final predicted output
print("Predicted Output after Training:")
print(predicted_output)
```
