```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# Input dataset (XOR problem)
X = np.array([[0, 0],
              [0, 1],
              [1, 0],
              [1, 1]])
# Expected output
y = np.array([[0], [1], [1], [0]])
# Create a Sequential neural network model
model = Sequential()
# Add hidden layer with 2 neurons and sigmoid activation
model.add(Dense(2, input_dim=2, activation='sigmoid'))
# Add output layer with 1 neuron
model.add(Dense(1, activation='sigmoid'))
# Compile the model
# loss: error calculation
# optimizer: updates weights using backpropagation
model.compile(loss='binary_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
# Train the model
# epochs: number of forward and backward passes
model.fit(X, y, epochs=5000, verbose=0)
# Predict output using trained model
predictions = model.predict(X)
# Display predicted values
print("Predicted Output:")
print(predictions)
```
